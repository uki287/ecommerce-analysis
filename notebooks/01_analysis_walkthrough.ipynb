# Olist 分析起点

这个 notebook 只做两件事：**加载已生成的图表底表**、**展示分析结果**。  
图表请在本 notebook 后面的单元格中自行添加；数据清洗和指标计算已经在 `src/` 中完成。

先运行 `python src/run_analysis.py` 生成 `report/tables/`。

```python
from pathlib import Path
import pandas as pd

pd.set_option("display.max_columns", 60)
pd.set_option("display.width", 160)

ROOT = Path.cwd()
while not (ROOT / "report" / "tables").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent

TABLES = ROOT / "report" / "tables"

def read_table(name: str) -> pd.DataFrame:
    return pd.read_csv(TABLES / f"{name}.csv")

print(ROOT)
print(len(list(TABLES.glob("*.csv"))), "tables")
```

```python
overview = read_table("overview_kpis")
overview.T
```

```python
monthly = read_table("monthly_kpis")
monthly[[
    "month", "orders_valid", "gmv", "gross_order_value", "aov",
    "customers", "repeat_order_rate", "late_rate", "review_score",
]]
```
